```python
from langchain_community.document_loaders import PyMuPDFLoader

file_path = "./app/data/LAND ACT, 2020 (ACT 1036).pdf"
loader = PyMuPDFLoader(file_path)
```

```python
docs = loader.load()
docs[0].page_content[:500]  # display first 500 characters of the first page
```
